```python
# Problem 1
# Apply all the preprocessing techniques that you think are necessary
```

```python
from bs4 import BeautifulSoup
def remove_html_tags(text):
    return BeautifulSoup(text,"html.parser").get_text(separator=" ")
```

```python
import re

def remove_urls(text):
 return re.sub(r'https?://\S+|www\.\S+','',text)
```

```python
def remove_backsalsh(text):
    return text.replace("\n"," ")
```

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm", disable=["ner", "parser"])
```

```python
def preprocessing_pipeline(text):
    doc = nlp(text)

    tokens = [token.lemma_.lower() for token in doc if not token.is_stop and not token.is_punct]

    return tokens

```

```python
df['review'] = df['review'].apply(remove_html_tags)
```

```python
df['review'] = df['review'].apply(remove_urls)
```

```python
df['review'] = df['review'].apply(remove_backsalsh)
```

```python
df['review'] = df['review'].apply(preprocessing_pipeline)
```

```python
# Problem 3
# Apply One Hot Encoding
```

```python
import numpy as np
```

```python
word_index = {word: i for i, word in enumerate(vocabulary)}
```

```python
def ohe(word):
  token= np.zeros(vocab_size,dtype=int)
  index = word_index[word]
  token[index] = 1
  return token
```

```python
one_hot_encoding = df['review'].apply(lambda row: [ ohe(token) for token in row])
```
